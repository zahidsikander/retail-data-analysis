```python

duplicates = df[df.duplicated(keep=False)]

print("Rows involved in duplicates:", len(duplicates))
display(duplicates.head(10))
```

```python

print("Rows before cleaning:", len(df))

df_clean = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df_clean))
print("Duplicate rows remaining:", df_clean.duplicated().sum())
print("Rows removed:", len(df) - len(df_clean))
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python

missing_summary = pd.DataFrame({
    "Missing Count": df_clean.isna().sum(),
    "Missing Percentage": (df_clean.isna().mean() * 100).round(2)
})

display(missing_summary)
```

```python

before = len(df_clean)

df_clean = df_clean.dropna(subset=["Description"]).copy()

print("Rows before:", before)
print("Rows after:", len(df_clean))
print("Rows removed:", before - len(df_clean))
print("Missing descriptions remaining:", df_clean["Description"].isna().sum())
print("Missing customer IDs remaining:", df_clean["CustomerID"].isna().sum())
```

```python

df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

print("InvoiceDate data type:", df_clean["InvoiceDate"].dtype)
print("Invalid dates:", df_clean["InvoiceDate"].isna().sum())

display(df_clean[["InvoiceNo", "InvoiceDate"]].head())

```

```python

print("Negative quantity rows:", (df_clean["Quantity"] < 0).sum())
print("Zero quantity rows:", (df_clean["Quantity"] == 0).sum())

print("Negative unit price rows:", (df_clean["UnitPrice"] < 0).sum())
print("Zero unit price rows:", (df_clean["UnitPrice"] == 0).sum())

print("\nSample rows with negative quantities:")
display(df_clean[df_clean["Quantity"] < 0].head(10))

print("\nSample rows with negative unit prices:")
display(df_clean[df_clean["UnitPrice"] < 0].head(10))
```

```python

zero_price = df_clean[df_clean["UnitPrice"] == 0]

print("Zero-price rows:", len(zero_price))
print("\nMost common zero-price product descriptions:")
display(zero_price["Description"].value_counts().head(15))

print("\nSample zero-price transactions:")
display(
    zero_price[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "CustomerID", "Country"]
    ].head(10)
)
```

```python

# Remove negative-price accounting adjustments
df_clean = df_clean[df_clean["UnitPrice"] >= 0].copy()

# Calculate transaction value
df_clean["Sales"] = df_clean["Quantity"] * df_clean["UnitPrice"]

# Save cleaned dataset
df_clean.to_csv("../outputs/cleaned_online_retail.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final rows:", len(df_clean))
print("Final columns:", len(df_clean.columns))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Group sales by month
monthly_sales = (
    df_clean.set_index("InvoiceDate")["Sales"]
    .resample("MS")
    .sum()
)

# Plot
plt.figure(figsize=(12, 5))
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

top_products = (
    df_clean[df_clean["Quantity"] > 0]
    .groupby("Description")["Quantity"]
    .sum()
    .nlargest(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))
top_products.plot(kind="barh")

plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python

country_sales = (
    df_clean.groupby("Country")["Sales"]
    .sum()
    .nlargest(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))
country_sales.plot(kind="barh")

plt.title("Top 10 Countries by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```
